# **BEAUTIFUL SOUP**
**Documentación:** https://beautiful-soup-4.readthedocs.io/en/latest/

```python
from bs4 import BeautifulSoup
import requests

# EMPEZAMOS CON EL ESCRAPING
# 1. Obtener el HTML
URL_BASE = 'https://scrapepark.org/courses/spanish/'
pedido_obtenido = requests.get(URL_BASE)
html_obtenido = pedido_obtenido.text

# 2. Parsear ese HTML
soup = BeautifulSoup(html_obtenido, 'html.parser')
```

**Ahora empezamos a inspeccionar la página web que queramos escrapear**

`find()`
Va a traer solo el primer elemento de esa etiqueta


```python
# Buscamos el primer H2
primer_h2 = soup.find('h2')
print(primer_h2)
```

`text`

```python
# Si sólo queremos el texto sin la etiqueta
primer_h2 = soup.h2.text
print(primer_h2)
```

`find_all()` Va a traer todos los elementos de esa etiqueta. Cuando existen más de 1 elemento, nos genera un objeto iterable.
`limit` Como segundo argumento, poner un limite de elementos de ese tipo que queremos que devuela

```python
# Traer todos los H2 de la URL
print('------------TRAER EL PRIMER H2-------------')
todos_h2 = soup.find_all('h2')
print(todos_h2)

# Traer todos los H2 de la URL con límite
print('------------TODOS LOS H2 CON LIMITE DE 1-------------')
todos_h2_limite = soup.find_all('h2', limit=1)
print(todos_h2_limite)

# Iteramos sobre el objeto
print('------------OBJETO-------------')
for s in todos_h2:
    print(s.text)
```

**Encontrar atributos en el HTML** (ID y Class)

```python
# CLASS
# class_ = Para los atributos de clase, especificar con un _
divs = soup.find_all('div', class_ = "heading-container heading-center")

for div in divs:
    print(div)
    print(' ')
```

**Descargar imágenes**

```python
# Todas las etiquetas que tengan el atributo "src"
src_all = soup.find_all(src=True)

for elemento in src_all:
    if elemento['src'].endswith('.jpg'):
        print(elemento)
```

```python
# Descargar TODAS las imágenes
url_imagenes = []

for i, imagen in enumerate(src_all):
    if imagen['src'].endswith('png'):
        print(imagen['src'])
        r = requests.get(f'https://scrapepark.org/courses/spanish/{imagen['src']}')
        
        with open(f'imagen_{i}.png', 'wb') as f:
            f.write(r.content)
```

**Etiquetas IFRAME y TABLE**

```python
# Información de tablas
URL_BASE = "https://scrapepark.org/courses/spanish"
URL_TABLA = soup.find_all('iframe')[0]['src']

request_tabla = requests.get(f'{URL_BASE}/{URL_TABLA}')
html_tabla = request_tabla.text
soup_tabla = BeautifulSoup(html_tabla, "html.parser")
soup_tabla.find('table')

productos_faltantes = soup_tabla.find_all(['th', 'td'], attrs={'style':'color: red;'})
productos_faltantes = [talle.text for talle in productos_faltantes]

print(productos_faltantes)
```

```python
# Obtener los productos con su precio
# Buscar todos los divs que tengan la clase detail-box
divs = soup.find_all('div', class_='detail-box')

# Creamos 2 listas para guardar el nombre del producto y el precio
productos = []
precios = []

for div in divs:
    if (div.h6 is not None) and ('Patineta' in div.h5.text):
        # Obtener el texto del h5 sin espacios
        producto = div.h5.get_text(strip=True)
        # Obtener el texto del h6 sin espacios y cambiamos el signo $ por el espacio
        precio = div.h6.get_text(strip=True).replace('$', '')
        # Se puede agregar filtros
        print(f'Producto: {producto} | Precio: {precio}')
        productos.append(producto)
        precios.append(precio)

```

**Cambios en las URLs dinámicas**

```python
""" 
La web tiene 2 páginas de contacto:
- https://scrapepark.org/courses/spanish/contact1
- https://scrapepark.org/courses/spanish/contact2
Se puede ver que solamente cambia el número de atrás. Entonces desde nuestra URL_BASE podemos
recorrer en el rango 1 y 2 para agregarle el valor al final
"""
URL_BASE = 'https://scrapepark.org/courses/spanish/contact'

for i in range(1,3):
    URL_FINAL = f'{URL_BASE}{i}'
    print(URL_FINAL)
    r = requests.get(URL_FINAL)
    soup = BeautifulSoup(r.text, 'html.parser')
    # Obtener el h5 que es el texto diferente en cada página de contacto
    print(soup.h5.text)
```

```python
# OBTENER EL NUMERO DE TELEFONO DE UN DETERMINADO PAIS
# Expresiones regulares
import re

# 1. Obtener el HTML
URL_BASE = 'https://scrapepark.org/courses/spanish/'
pedido_obtenido = requests.get(URL_BASE)
html_obtenido = pedido_obtenido.text

# 2. Parsear ese HTML
soup = BeautifulSoup(html_obtenido, 'html.parser')

# Le pasamos un argumento string y usar dentro la libreria RE para expresiones regulares
# con el método .compile()
telefonos = soup.find_all(string=re.compile('\\d+-\\d+-\\d+'))
telefonos
```

**Text vs get_text** <br>
`parent`Nos posiciona al elemento sobre la etiqueta que lo contiene <br>
`parent.next_sibilings` Obtenemos todos los elementos que están al mismo nivel

```python
copyrights = soup.find_all(string=re.compile('©'))
print(copyrights[0])

# Trae todo lo que hay en la etiqueta padre
primer_copyright = copyrights[0]
print(primer_copyright.parent)

# Ejemplo con elementos al mismo nivel
menu = soup.find(string=re.compile('MENÚ'))
print(f'MISMO NIVEL: {menu.parent}')

# Obtenemos todos los elementos que están al mismo nivel
print(f'SIBLINGS: {menu.parent.find_next_siblings()}')
```

**Almacenamiento de los datos `CSV`**

```python
import csv

productos.insert(0, 'productos')
precios.insert(0, 'precios')
"""
- Construimos un diccionario.
- Zip nos permite combinar 2 o más listas por posición en forma de tupla y las convierte en un diciconario:
- Ejemplo: une cada producto con su precio correspondiente.
"('productos', 'precios')","('Patineta Nueva 1', '75')","('Patineta Usada 2', '80')","('Patineta Nueva 3', '68')"
"""
datos = dict(zip(productos, precios))
with open('datos.csv', 'w') as f:
    w = csv.writer(f)
    w.writerow(datos.items())
```

3. Traer cualquier texto de la pagina que tenga la palabra descuento u oferta.
4. Generar un archivo. csv con dos columnas: Una conteniendo el nombre del cliente y otra su testimonio.

```python
# Obtener las patinetas que salgan menos que $68
patinetas = soup.find_all("div", class_="detail-box")

patineta_nombre = []
patineta_precio = []

for p in patinetas:
    if p.h5 and p.h6:
        if ("Patineta" in p.h5.text):
            producto = p.h5.get_text(strip=True)
            precio = int(p.h6.get_text(strip=True).replace("$", ""))
            
            if precio < 68:
                print(f"PRODUCTO: {producto} | PRECIO: {precio}")
                patineta_nombre.append(producto)
                patineta_precio.append(precio)
```

```python
# Obtener las patinetas que en su nombre tengan un numero mayor a 3
patinetas = soup.find_all('div', class_='detail-box')
patineta_nombre = []

for p in patinetas:
    if p.h5:
        if ('Patineta' in p.h5.text):
            patineta = p.h5.get_text(strip=True)
            patineta_split = int(patineta.split()[2])
            
            if patineta_split > 3:
                patineta_nombre.append(patineta)
                print(patineta)
```
