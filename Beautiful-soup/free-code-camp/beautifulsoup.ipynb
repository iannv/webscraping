# **BEAUTIFUL SOUP**
**Documentación:** https://beautiful-soup-4.readthedocs.io/en/latest/

```python
from bs4 import BeautifulSoup
import requests

# EMPEZAMOS CON EL ESCRAPING
# 1. Obtener el HTML
URL_BASE = 'https://scrapepark.org/courses/spanish/'
pedido_obtenido = requests.get(URL_BASE)
html_obtenido = pedido_obtenido.text

# 2. Parsear ese HTML
soup = BeautifulSoup(html_obtenido, 'html.parser')
```

**Ahora empezamos a inspeccionar la página web que queramos escrapear**

`find()`
Va a traer solo el primer elemento de esa etiqueta


```python
# Buscamos el primer H2
primer_h2 = soup.find('h2')
print(primer_h2)
```

`text`

```python
# Si sólo queremos el texto sin la etiqueta
primer_h2 = soup.h2.text
print(primer_h2)
```

`find_all()` Va a traer todos los elementos de esa etiqueta. Cuando existen más de 1 elemento, nos genera un objeto iterable.
`limit` Como segundo argumento, poner un limite de elementos de ese tipo que queremos que devuela

```python
# Traer todos los H2 de la URL
print('------------TRAER EL PRIMER H2-------------')
todos_h2 = soup.find_all('h2')
print(todos_h2)

# Traer todos los H2 de la URL con límite
print('------------TODOS LOS H2 CON LIMITE DE 1-------------')
todos_h2_limite = soup.find_all('h2', limit=1)
print(todos_h2_limite)

# Iteramos sobre el objeto
print('------------OBJETO-------------')
for s in todos_h2:
    print(s.text)
```

**Encontrar atributos en el HTML** (ID y Class)

```python
# CLASS
# class_ = Para los atributos de clase, especificar con un _
divs = soup.find_all('div', class_ = "heading-container heading-center")

for div in divs:
    print(div)
    print(' ')
```

**Descargar imágenes**

```python
# Todas las etiquetas que tengan el atributo "src"
src_all = soup.find_all(src=True)

for elemento in src_all:
    if elemento['src'].endswith('.jpg'):
        print(elemento)
```

```python
# Descargar TODAS las imágenes
url_imagenes = []

for i, imagen in enumerate(src_all):
    if imagen['src'].endswith('png'):
        print(imagen['src'])
        r = requests.get(f'https://scrapepark.org/courses/spanish/{imagen['src']}')
        
        with open(f'imagen_{i}.png', 'wb') as f:
            f.write(r.content)
```

**Etiquetas IFRAME y TABLE**

```python
# Información de tablas
URL_BASE = "https://scrapepark.org/courses/spanish"
URL_TABLA = soup.find_all('iframe')[0]['src']

request_tabla = requests.get(f'{URL_BASE}/{URL_TABLA}')
html_tabla = request_tabla.text
soup_tabla = BeautifulSoup(html_tabla, "html.parser")
soup_tabla.find('table')

productos_faltantes = soup_tabla.find_all(['th', 'td'], attrs={'style':'color: red;'})
productos_faltantes = [talle.text for talle in productos_faltantes]

print(productos_faltantes)
```
